Calculate confidence intervals for critical business variables, using bootstrapping and parametric approaches to quantify the uncertainty of estimates and support reliable forecasting and planning.

```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
np.random.seed(42)
```

```python
daily_revenue = np.random.exponential(scale=5000, size=100) + 10000
```

```python
df = pd.DataFrame({'revenue': daily_revenue})
```

```python
print("First few rows of revenue data:")
print(df.head())
print("\nSample Mean Revenue:", round(df['revenue'].mean(), 2))
```

```python
sample_mean = df['revenue'].mean()
sample_std = df['revenue'].std(ddof=1)
n = len(df['revenue'])
```

```python
confidence_level = 0.95
ci_parametric = stats.t.interval(
    confidence=confidence_level,
    df=n-1,
    loc=sample_mean,
    scale=sample_std / np.sqrt(n)
)
```

```python
print("--- Parametric 95% Confidence Interval ---")
print("Lower Bound:", round(ci_parametric[0], 2))
print("Upper Bound:", round(ci_parametric[1], 2))
```

```python
n_bootstraps = 1000
boot_means = []
```

```python
for _ in range(n_bootstraps):
    # Draw random sample with replacement
    boot_sample = np.random.choice(df['revenue'], size=n, replace=True)
    boot_means.append(np.mean(boot_sample))
```

```python
ci_lower_boot = np.percentile(boot_means, 2.5)
ci_upper_boot = np.percentile(boot_means, 97.5)
```

```python
print("--- Bootstrapped 95% Confidence Interval ---")
print("Lower Bound:", round(ci_lower_boot, 2))
print("Upper Bound:", round(ci_upper_boot, 2))
```

```python
comparison = pd.DataFrame({
    'Method': ['Parametric (T-Test)', 'Bootstrapping'],
    'Mean': [sample_mean, np.mean(boot_means)],
    'Lower 95% Bound': [ci_parametric[0], ci_lower_boot],
    'Upper 95% Bound': [ci_parametric[1], ci_upper_boot]
})
```

```python
print("Confidence Interval Comparison Table:")
print(comparison.round(2))
```
