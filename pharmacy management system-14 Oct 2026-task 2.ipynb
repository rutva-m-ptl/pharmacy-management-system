Conduct hypothesis testing across multiple business scenarios by formulating null and alternative hypotheses, selecting appropriate statistical tests (t-test, z-test), and interpreting p-values to drive data-driven decision-making.

```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
np.random.seed(42)
```

```python
old_page_sales = np.random.normal(loc=120, scale=15, size=100)
new_page_sales = np.random.normal(loc=126, scale=15, size=100)
```

```python
time_before_training = np.random.normal(loc=45, scale=8, size=50)
time_after_training = time_before_training - np.random.normal(loc=4, scale=3, size=50)
```

```python
df_ab = pd.DataFrame({'old_design': old_page_sales, 'new_design': new_page_sales})
df_train = pd.DataFrame({'before': time_before_training, 'after': time_after_training})
```

```python
print("Website Sales Sample (A/B Test):")
print(df_ab.head())
```

```python
alpha = 0.05
```

```python
t_stat, p_value = stats.ttest_ind(df_ab['new_design'], df_ab['old_design'])
```

```python
print("--- Scenario 1: A/B Test Results ---")
print("T-Statistic:", round(t_stat, 4))
print("P-Value:", round(p_value, 4))
```

```python
if p_value < alpha:
    print("Decision: Reject Null Hypothesis (H0). New design significantly changed sales!")
else:
    print("Decision: Fail to Reject Null Hypothesis (H0). No significant difference found.")
```

```python
t_stat_paired, p_value_paired = stats.ttest_rel(df_train['before'], df_train['after'])
```

```python
print("--- Scenario 2: Paired Test Results ---")
print("T-Statistic:", round(t_stat_paired, 4))
print("P-Value:", round(p_value_paired, 4))
```

```python
if p_value_paired < alpha:
    print("Decision: Reject Null Hypothesis (H0). Training significantly reduced response time!")
else:
    print("Decision: Fail to Reject Null Hypothesis (H0). Training had no clear effect.")
```

```python
from statsmodels.stats.proportion import proportions_ztest
```

```python
clicks = np.array([120, 160]) 
views = np.array([1000, 1000])
```

```python
z_stat, p_val_prop = proportions_ztest(count=clicks, nobs=views)
```

```python
print("--- Scenario 3: Z-Test for Proportions Results ---")
print("Z-Statistic:", round(z_stat, 4))
print("P-Value:", round(p_val_prop, 4))
```

```python
if p_val_prop < alpha:
    print("Decision: Reject Null Hypothesis (H0). Campaign B click rate is significantly higher!")
else:
    print("Decision: Fail to Reject Null Hypothesis (H0). Both campaigns perform similarly.")
```
