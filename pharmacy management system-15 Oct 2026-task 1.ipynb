Perform chi-square tests of independence to explore and interpret relationships between categorical variables, evaluating statistical significance and guiding feature selection decisions.

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
```

```python
membership_types = ['Basic', 'Silver', 'Gold']
churn_status = ['No', 'Yes']
```

```python
df = pd.DataFrame({
    'membership': np.random.choice(membership_types, size=300, p=[0.5, 0.3, 0.2]),
    'churn': np.random.choice(churn_status, size=300, p=[0.7, 0.3])
})
```

```python
print("First few rows of categorical data:")
print(df.head())
```

```python
contingency_table = pd.crosstab(df['membership'], df['churn'])
```

```python
print("Contingency Table (Observed Counts):")
print(contingency_table)
```

```python
from scipy.stats import chi2_contingency
```

```python
chi2_stat, p_value, dof, expected_counts = chi2_contingency(contingency_table)
```

```python
print("--- Chi-Square Test Results ---")
print("Chi2 Statistic:", round(chi2_stat, 4))
print("P-Value:", round(p_value, 4))
print("Degrees of Freedom:", dof)
```

```python
alpha = 0.05
```

```python
print("--- Decision & Interpretation ---")
if p_value < alpha:
    print("Decision: Reject Null Hypothesis (H0).")
    print("Interpretation: There is a statistically significant relationship between Membership and Churn.")
    print("Feature Selection Action: KEEP 'membership' as a feature in your model.")
else:
    print("Decision: Fail to Reject Null Hypothesis (H0).")
    print("Interpretation: Membership level and Churn are independent of each other.")
    print("Feature Selection Action: DROP 'membership' or reconsider how it is encoded.")
```
