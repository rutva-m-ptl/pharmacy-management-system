<h2>Normalize and encode inconsistent categorical variables by standardizing case sensitivity, resolving variations, mapping synonyms, and applying appropriate encoding techniques like one-hot, label encoding, or frequency encoding for machine learning readin</h2>

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    'payment_method': [' cash ', 'CASH', 'Credit Card', 'credit_card', 'cc', 'UPI', 'upi ', ' Debit Card '],
    'status': ['Completed', 'completed', 'DONE', 'pending', 'Pending', 'CANCELLED', 'cancelled', 'Completed'],
    'category': ['urinary tract', 'skin infections', 'acne', 'acne', 'urinary tract', 'acne', 'skin infections', 'acne']
})
```

```python
print("Original Data:")
print(df)
```

```python
categorical_cols = ['payment_method', 'status', 'category']
```

```python
for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()
```

```python
print("After Whitespace Removal & Lowercasing:")
print(df[['payment_method', 'status']])
```

```python
payment_synonyms = {
    'cash': 'cash',
    'credit card': 'credit_card',
    'credit_card': 'credit_card',
    'cc': 'credit_card',
    'debit card': 'debit_card',
    'upi': 'upi'
}
```

```python
status_synonyms = {
    'completed': 'completed',
    'done': 'completed',
    'pending': 'pending',
    'cancelled': 'cancelled'
}
```

```python
df['payment_method'] = df['payment_method'].map(payment_synonyms)
df['status'] = df['status'].map(status_synonyms)
```

```python
print("After Synonym Resolution:")
print(df[['payment_method', 'status']])
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
status_order = {'cancelled': 0, 'pending': 1, 'completed': 2}
df['status_ordinal_encoded'] = df['status'].map(status_order)
```

```python
category_freq = df['category'].value_counts(normalize=True).to_dict()
df['category_freq_encoded'] = df['category'].map(category_freq)
```

```python
df_ohe = pd.get_dummies(df['payment_method'], prefix='payment', dtype=int)
```

```python
df_final = pd.concat([df, df_ohe], axis=1)
```

```python
print("Final Preprocessed & Encoded DataFrame:")
df_final
```
