<h2>Apply dimensionality reduction techniques like Principal Component Analysis (PCA), ensuring optimal component selection based on explained variance ratio and interpreting principal components in relation to original features.</h2>

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
data = {
    'height': np.random.normal(170, 10, 100),
    'weight': np.random.normal(70, 15, 100),
    'shoe_size': np.random.normal(40, 3, 100),
    'age': np.random.normal(30, 5, 100),
    'income': np.random.normal(50000, 10000, 100)
}
```

```python
df = pd.DataFrame(data)
```

```python
print(df.head())
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaler = StandardScaler()
```

```python
scaled_data = scaler.fit_transform(df)
```

```python
df_scaled = pd.DataFrame(scaled_data, columns=df.columns)
```

```python
print(df_scaled.head())
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA()
pca.fit(df_scaled)
```

```python
explained_variance = pca.explained_variance_ratio_
```

```python
print("Variance kept by each component:")
for i, var in enumerate(explained_variance):
    print("Component", i+1, ":", round(var * 100, 2), "%")
```

```python
print("\nTotal info kept if we pick first 2 components:", 
      round(sum(explained_variance[:2]) * 100, 2), "%")
```

```python
pca_final = PCA(n_components=2)
```

```python
pca_result = pca_final.fit_transform(df_scaled)
```

```python
df_pca = pd.DataFrame(pca_result, columns=['PC1', 'PC2'])
```

```python
weights = pd.DataFrame(pca_final.components_, columns=df.columns, index=['PC1', 'PC2'])
```

```python
print("How original features connect to PC1 and PC2:")
print(weights)
```
