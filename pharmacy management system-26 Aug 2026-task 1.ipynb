<h2>Apply complex log, Box-Cox, and Yeo-Johnson transformations to skewed numerical variables, ensuring improved data symmetry and validating transformations through distribution analysis and model performance improvements.</h2>

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
```

```python
skewed_data = np.random.exponential(scale=2.0, size=200)
```

```python
df = pd.DataFrame({'income': skewed_data})
```

```python
y = df['income'] * 5000 + np.random.normal(0, 1000, 200)
```

```python
print("First few rows of raw data:")
print(df.head())
```

```python
print("Original Skewness:", df['income'].skew())
```

```python
from scipy import stats
```

```python
df['log_income'] = np.log1p(df['income'])
```

```python
df['boxcox_income'], _ = stats.boxcox(df['income'] + 1e-5)
```

```python
df['yeojohnson_income'], _ = stats.yeojohnson(df['income'])
```

```python
print("Data after applying all transformations:")
print(df.head())
```

```python
print("Skewness Scores (Closer to 0 is better):")
print("Original:", round(df['income'].skew(), 3))
print("Log:", round(df['log_income'].skew(), 3))
print("Box-Cox:", round(df['boxcox_income'].skew(), 3))
print("Yeo-Johnson:", round(df['yeojohnson_income'].skew(), 3))
```

```python
from sklearn.linear_model import LinearRegression
```

```python
from sklearn.model_selection import cross_val_score
```

```python
model = LinearRegression()
```

```python
score_orig = cross_val_score(model, df[['income']], y, cv=5).mean()
```

```python
score_log = cross_val_score(model, df[['log_income']], y, cv=5).mean()
```

```python
score_yj = cross_val_score(model, df[['yeojohnson_income']], y, cv=5).mean()
```

```python
print("Model R2 Scores (Higher is better):")
print("Original Feature Score:", round(score_orig, 4))
print("Log Feature Score:", round(score_log, 4))
print("Yeo-Johnson Feature Score:", round(score_yj, 4))
```
