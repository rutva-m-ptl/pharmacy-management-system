Analyze multicollinearity among predictors using Variance Inflation Factor (VIF), identifying highly correlated variables, and applying dimensionality reduction or regularization techniques to mitigate redundancy.

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
```

```python
x1 = np.random.uniform(10, 100, 200)               
x2 = x1 * 2 + np.random.normal(0, 0.5, 200)        
x3 = np.random.uniform(1, 10, 200)
```

```python
y = 5 + (1.5 * x1) + (0.8 * x3) + np.random.normal(0, 2, 200)
```

```python
df = pd.DataFrame({'feature_1': x1, 'feature_2': x2, 'feature_3': x3})
```

```python
print("First few rows of dataset:")
print(df.head())
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
vif_data = pd.DataFrame()
vif_data['Feature'] = df.columns
vif_data['VIF'] = [variance_inflation_factor(df.values, i) for i in range(df.shape[1])]
```

```python
print("VIF Scores (High VIF > 10 means feature is redundant):")
print(vif_data)
```

```python
df_reduced = df.drop(columns=['feature_2'])
```

```python
vif_reduced = pd.DataFrame()
vif_reduced['Feature'] = df_reduced.columns
vif_reduced['VIF'] = [variance_inflation_factor(df_reduced.values, i) for i in range(df_reduced.shape[1])]
```

```python
print("VIF Scores after dropping feature_2:")
print(vif_reduced)
```

```python
from sklearn.linear_model import Ridge, LinearRegression
from sklearn.model_selection import cross_val_score
```

```python
lin_reg = LinearRegression()
ridge_reg = Ridge(alpha=10.0)
```

```python
score_lin = cross_val_score(lin_reg, df, y, cv=5).mean()
score_ridge = cross_val_score(ridge_reg, df, y, cv=5).mean()
```

```python
print("Cross-Validation R2 Scores:")
print("Standard Linear Regression Score:", round(score_lin, 4))
print("Ridge Regression Score:", round(score_ridge, 4))
```
