<h2>Generate polynomial and interaction features to model non-linear relationships in the dataset, performing regularization techniques (Ridge, Lasso) to mitigate multicollinearity introduced by these additional features.</h2>

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
```

```python
X1 = np.random.uniform(1, 10, 100)
```

```python
X2 = np.random.uniform(1, 5, 100)
```

```python
y = 10 + (2 * X1) + (0.5 * (X1 ** 2)) + (3 * X1 * X2) + np.random.normal(0, 2, 100)
```

```python
df = pd.DataFrame({'size': X1, 'bedrooms': X2})
```

```python
print(df.head())
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly = PolynomialFeatures(degree=2, include_bias=False)
```

```python
X_poly = poly.fit_transform(df)
```

```python
feature_names = poly.get_feature_names_out(df.columns)
```

```python
df_poly = pd.DataFrame(X_poly, columns=feature_names)
```

```python
print("New dataset with squared and mixed features:")
print(df_poly.head())
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
X_scaled = scaler.fit_transform(df_poly)
```

```python
from sklearn.linear_model import Ridge, Lasso
```

```python
ridge = Ridge(alpha=1.0)
```

```python
ridge.fit(X_scaled, y)
```

```python
lasso = Lasso(alpha=0.5)
```

```python
lasso.fit(X_scaled, y)
```

```python
results = pd.DataFrame({
    'Feature': feature_names,
    'Ridge_Weights': ridge.coef_,
    'Lasso_Weights': lasso.coef_
})
```

```python
print("Weights for each feature:")
print(results)
```
