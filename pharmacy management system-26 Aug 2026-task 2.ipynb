<h2>Engineer sophisticated time-dependent features using moving averages, exponential smoothing, and rolling statistics, followed by trend and seasonality decomposition to uncover actionable insights in temporal datasets.</h2>

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
```

```python
dates = pd.date_range(start='2026-01-01', periods=100, freq='D')
```

```python
trend = np.linspace(10, 50, 100)
season = 5 * np.sin(np.linspace(0, 20, 100))
noise = np.random.normal(0, 2, 100)
```

```python
df = pd.DataFrame({
    'date': dates,
    'sales': trend + season + noise
}).set_index('date')
```

```python
print("First few rows of sales data:")
print(df.head())
```

```python
df['moving_avg_7'] = df['sales'].shift(1).rolling(window=7).mean()
```

```python
df['rolling_std_7'] = df['sales'].shift(1).rolling(window=7).std()
```

```python
df['ewma_7'] = df['sales'].shift(1).ewm(span=7).mean()
```

```python
print("Data with time features added:")
print(df.dropna().head())
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
df_clean = df.dropna().copy()
```

```python
decomp = seasonal_decompose(df_clean['sales'], model='additive', period=7)
```

```python
df_clean['trend'] = decomp.trend
```

```python
df_clean['seasonal'] = decomp.seasonal
```

```python
df_clean['leftover_noise'] = decomp.resid
```

```python
print("Decomposed parts of the time series:")
print(df_clean[['sales', 'trend', 'seasonal', 'leftover_noise']].head(10))
```
